```python
from base64 import b64encode
from pathlib import Path
from IPython.display import HTML, display

photo_folder = Path(".")
def embed_photo(filename, alt, class_name):
    image_data = b64encode((photo_folder / filename).read_bytes()).decode("ascii")
    return f'<figure class="memory-photo {class_name}"><img src="data:image/png;base64,{image_data}" alt="{alt}"></figure>'

photo_path = Path("photo_5453862477801583418_y копия.png")
photo_data = b64encode(photo_path.read_bytes()).decode("ascii")
photo_tag = f'<img class="dasha-photo" src="data:image/png;base64,{photo_data}" alt="Даша в детстве">'
wrapped_photo_tag = f'<span class="dasha-photo-bg">{photo_tag}</span>'
photo_together = embed_photo("3.png", "Даша с подругой", "photo-together")
photo_braids = embed_photo("4.png", "Даша с косичками", "photo-braids")
photo_guinea = embed_photo("2.png", "Даша с морской свинкой", "photo-guinea")
photo_headband = embed_photo("5.png", "Даша в праздничном образе", "photo-headband")

fortune_html = '''
    <section class="fortune-section" aria-labelledby="fortune-title">
      <div class="section-head"><span class="section-number">06</span><h2 id="fortune-title">Колесо фортуны</h2></div>
      <p class="fortune-intro">Крути колесо два раза и посмотри, какие цифры выпадут!</p>
      <div class="wheel-stage">
        <div class="wheel-pointer" aria-hidden="true"></div>
        <button class="fortune-wheel" id="fortune-wheel" type="button" aria-label="Крутить колесо">
          <span class="wheel-number number-one">1</span><span class="wheel-number number-two">2</span>
          <span class="wheel-number number-three">3</span><span class="wheel-number number-four">4</span><span class="wheel-center">♡</span>
        </button>
      </div>
      <button class="spin-button" id="spin-button" type="button">Крутить колесо</button>
      <p class="spin-status" id="spin-status">Осталось вращений: 2</p>
      <p class="spin-results" id="spin-results">Выпавшие цифры: <span>пока нет</span></p>
    </section>
'''
fortune_css = '''
.fortune-section { margin: 92px clamp(4px, 5vw, 60px) 0; text-align: center; }
.fortune-section .section-head { text-align: left; }
.fortune-section h2 { color: #90A63F; }
.fortune-intro, .spin-status, .spin-results { color: #90A63F; font-size: 14px; line-height: 1.6; }
.wheel-stage { position: relative; width: min(280px, 72vw); margin: 34px auto 24px; }
.fortune-wheel { position: relative; display: block; width: 100%; aspect-ratio: 1; margin: 0; padding: 0; border: 10px solid #A63F96; border-radius: 50%; background: conic-gradient(#F5FFD1 0 25%, #A63F96 25% 50%, #90A63F 50% 75%, #ffc2f5 75% 100%); box-shadow: none; transform: rotate(0deg); transition: transform 3.2s cubic-bezier(.17,.67,.12,1); }
.fortune-wheel:disabled { cursor: default; }
.wheel-number { position: absolute; left: 50%; top: 50%; color: #202e2a; font: 600 32px Fraunces, serif; }
.number-one { transform: translate(-50%, -120px); }
.number-two { transform: translate(70px, -50%); }
.number-three { transform: translate(-50%, 75px); }
.number-four { transform: translate(-100px, -50%); }
.wheel-center { position: absolute; left: 50%; top: 50%; display: grid; width: 56px; height: 56px; place-items: center; transform: translate(-50%, -50%); border: 5px solid #A63F96; border-radius: 50%; background: #F5FFD1; color: #A63F96; font-size: 27px; }
.wheel-pointer { position: absolute; z-index: 3; top: -8px; left: 50%; width: 0; height: 0; transform: translateX(-50%); border-right: 13px solid transparent; border-left: 13px solid transparent; border-top: 28px solid #A63F96; }
.spin-button { margin: 0 auto; background: #A63F96; }
.spin-button:disabled { opacity: .45; cursor: default; transform: none; }
.spin-results span { color: #A63F96; font-weight: 700; }
@media (max-width: 600px) { .fortune-section { margin-top: 68px; } .number-one { transform: translate(-50%, -100px); } .number-three { transform: translate(-50%, 60px); } .number-four { transform: translate(-80px, -50%); } .number-two { transform: translate(58px, -50%); } }
'''
fortune_script = '''
const fortuneWheel = document.getElementById('fortune-wheel');
const spinButton = document.getElementById('spin-button');
const spinStatus = document.getElementById('spin-status');
const spinResults = document.querySelector('#spin-results span');
const fortuneStorageKey = 'dasha-fortune-wheel-results';
let fortuneState = { spins: [], rotation: 0 };
try { fortuneState = JSON.parse(localStorage.getItem(fortuneStorageKey) || '{"spins":[],"rotation":0}'); } catch (error) {}
function renderFortuneState() {
  const remaining = Math.max(0, 2 - fortuneState.spins.length);
  spinStatus.textContent = `Осталось вращений: ${remaining}`;
  spinResults.textContent = fortuneState.spins.length ? fortuneState.spins.join(' · ') : 'пока нет';
  if (fortuneState.spins.length >= 2) { spinButton.disabled = true; fortuneWheel.disabled = true; spinButton.textContent = 'Попытки использованы'; }
}
function spinFortuneWheel() {
  if (fortuneState.spins.length >= 2) return;
  const result = Math.floor(Math.random() * 4) + 1;
  fortuneState.spins.push(result);
  fortuneState.rotation += 1440 + Math.floor(Math.random() * 360);
  fortuneWheel.style.transform = `rotate(${fortuneState.rotation}deg)`;
  localStorage.setItem(fortuneStorageKey, JSON.stringify(fortuneState));
  renderFortuneState();
}
spinButton.addEventListener('click', spinFortuneWheel);
fortuneWheel.addEventListener('click', spinFortuneWheel);
renderFortuneState();
'''

page = page.replace("background: #ffc2f5;", "background: #ffc2f5;")
page = page.replace("background: var(--sage);", "background: #F5FFD1;")
page = page.replace("color: var(--ink); font: 500 16px Manrope", "color: #A63F96; font: 500 16px Manrope")
page = page.replace("color: #98a49e;", "color: #A63F96;")
page = page.replace("color: var(--muted); font-size: 14px; line-height: 1.65;", "color: #90A63F; font-size: 14px; line-height: 1.65;")
page = page.replace("--line: #d8e2d8;", "--line: #F2FFC2;")
page = page.replace("#c75f50", "#90A63F")
page = page.replace("#397b80", "#A63F96")
page = page.replace("#aa7b18", "#90A63F")
page = page.replace("#537c60", "#A63F96")
page = page.replace("#795a90", "#F5FFD1")
if ".fortune-wheel" not in page:
    page = page.replace("</style>", f"{fortune_css}</style>", 1)
page = page.replace("box-shadow: 0 14px 28px rgba(32,46,42,.12);", "")
page = page.replace("filter: drop-shadow(0 10px 12px rgba(32,46,42,.16));", "filter: none;")
page = page.replace("background: #ffc2f5; filter: none;", "background: #F5FFD1; filter: none;", 1)
page = page.replace("Сегодня я хочу исправить то, что мы с тобой не делали по причине", "Сегодня я хочу исправить то, что мы с тобой не делали<br>по причине")
if '<span class="dasha-photo-bg">' not in page:
    if photo_tag in page:
        page = page.replace(photo_tag, wrapped_photo_tag, 1)
    else:
        page = page.replace('<section class="hero">', f'<section class="hero">{wrapped_photo_tag}', 1)
if '<figure class="memory-photo photo-together"' not in page:
    page = page.replace('    <section class="section"><div class="section-head"><span class="section-number">03</span>', f'    {photo_together}\n    <section class="section"><div class="section-head"><span class="section-number">03</span>', 1)
    page = page.replace('    <section class="section"><div class="section-head"><span class="section-number">04</span>', f'    {photo_braids}\n    <section class="section"><div class="section-head"><span class="section-number">04</span>', 1)
    page = page.replace('    <section class="section"><div class="section-head"><span class="section-number">05</span>', f'    {photo_guinea}\n    <section class="section"><div class="section-head"><span class="section-number">05</span>', 1)
    page = page.replace('    <section class="final">', f'    {photo_headband}\n    <section class="final">', 1)
if '<section class="fortune-section"' not in page:
    page = page.replace('    <section class="final">', fortune_html + '    <section class="final">', 1)
if 'const fortuneWheel' not in page:
    page = page.replace('</script>', fortune_script + '</script>', 1)
display(HTML(page))
```
